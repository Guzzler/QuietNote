## 2.1 The Discovery Framework (Adapted SPIN for AI Solutions)

The SPIN framework (Situation, Problem, Implication, Need-payoff) was developed for consultative selling, but it maps perfectly to AI solution design. The idea is simple: you move through four layers of understanding, each one deeper than the last.

### S -- Situation
*Establish the landscape. Understand what exists before proposing what should exist.*

| Question | Why It Matters |
|---|---|
| What does your organization do? | Sets context for domain, vocabulary, and risk tolerance |
| What's the current workflow for this task? | Reveals the baseline you're improving on |
| What tools/systems do you use today? | Identifies integration points and constraints |
| Who performs this task now, and how many people? | Reveals scale, cost, and expertise level |
| How long has this process been in place? | Signals whether they've tried to fix this before |

### P -- Problem
*Find the pain. The real pain, not the surface-level complaint.*

| Question | Why It Matters |
|---|---|
| What's painful about the current process? | Opens the door to the real problem |
| What takes too long? | Identifies latency/throughput bottlenecks |
| Where do errors happen? | Reveals quality gaps AI might address |
| What do your people wish they didn't have to do? | Points to the highest-value automation targets |
| What have you tried before? Why didn't it work? | Prevents you from re-proposing a failed approach |

### I -- Implication
*Quantify the cost of inaction. This is what makes the project worth doing.*

| Question | Why It Matters |
|---|---|
| What happens when this goes wrong? | Reveals severity -- annoying vs. catastrophic |
| What's the cost of the status quo? (time, money, morale) | Builds the business case |
| How does this affect downstream work? | Shows you think in systems, not silos |
| Are you losing customers/clients because of this? | Connects to revenue impact |
| What can't you do today because of this bottleneck? | Reveals aspirational goals |

### N -- Need-Payoff
*Anchor on outcomes. Get them to describe success in their own words.*

| Question | Why It Matters |
|---|---|
| If we could solve X, what would that mean for your team? | They sell themselves on the solution |
| What does success look like in 30 days? 6 months? | Defines scope for MVP vs. full rollout |
| What would make this worth the investment? | Reveals their real evaluation criteria |
| Who else benefits if this works? | Identifies stakeholders and champions |

---

**How to use SPIN in the interview**: You don't need to march through these in order like a checklist. Think of them as layers. Start with Situation, but weave in Problem and Implication questions as opportunities arise. Save Need-payoff for when you're ready to transition from discovery to solution design.

**Key principle**: Spend 60-70% of your discovery time in S and P. Most candidates rush through these to get to the "fun" part (building). The interviewers are watching for patience and thoroughness.

## 2.2 The Essential Question Bank

These are the questions you should have loaded and ready. Not all of them will apply in every scenario, but knowing the full set means you'll never be stuck wondering what to ask next.

---

### Understanding the Task

These questions define WHAT the system needs to do.

- **What exactly should the output look like? Can you show me an example?**  
  *This is the single most important question. An example output anchors everything. If they can't show you one, that's a signal the requirements aren't clear yet.*

- **How is this done today?** (manual, rule-based, other AI, outsourced?)  
  *Knowing the baseline tells you what "better" means. If humans do it, you inherit their quality bar. If a rules engine does it, you know there's some structure.*

- **What's the volume?** (10/day vs. 10,000/day)  
  *This massively affects architecture. 10/day = interactive, human-in-the-loop. 10,000/day = batch, automated, cost-sensitive.*

- **Who uses the output?** (internal team, customer-facing, regulatory?)  
  *Determines quality bar, tone, format requirements, and risk tolerance.*

- **Is this a one-time analysis or an ongoing workflow?**  
  *One-time = quick script. Ongoing = needs monitoring, error handling, maintenance.*

---

### Understanding the Data

These questions define WHAT the system works with.

- **What does the input data look like?** (format, length, language, structure)  
  *"Emails" could mean 3-sentence requests or 10-page threads with attachments.*

- **How variable is the input?** (structured vs. messy, consistent vs. varied)  
  *High variability = need more robust prompting, more examples, more testing.*

- **Are there edge cases I should know about?**  
  *There always are. This question gives them permission to share the weird stuff.*

- **Any sensitive or PII data?**  
  *Affects everything: deployment architecture, data handling, compliance, model choice.*

- **How much context is needed to do this task well?**  
  *Determines whether you need RAG, long context, or can work with single inputs.*

- **Do you have labeled examples?** (input-output pairs)  
  *If yes: few-shot prompting becomes easy. If no: you'll need to work harder on prompts.*

---

### Understanding Success

These questions define WHAT "good" means.

- **How do you evaluate quality today?**  
  *If they have an existing rubric or QA process, you can align to it. If they don't, you'll need to help them define one.*

- **What accuracy is acceptable?** (95%? 99%? 100%?)  
  *100% is a red flag -- it means either the task is trivial (use rules, not AI) or expectations need calibrating. 95% is usually a great starting point.*

- **What happens when the system gets it wrong?** (risk assessment)  
  *Wrong email routing = minor annoyance. Wrong medical classification = potential harm. This determines your guardrail strategy.*

- **What's the first milestone -- what would make this useful even if imperfect?**  
  *Identifies the MVP. Often, 80% of the value comes from solving the simple cases and flagging the hard ones for human review.*

- **How will you measure ROI?**  
  *Time saved? Errors reduced? Revenue gained? This shapes what you optimize for.*

---

### Understanding Constraints

These questions define the BOUNDARIES of the solution.

- **Latency requirements?** (real-time vs. batch, seconds vs. minutes)  
  *Real-time needs streaming, caching, maybe a smaller/faster model. Batch can use Opus and take its time.*

- **Budget considerations?**  
  *Affects model choice (Haiku vs. Sonnet vs. Opus), caching strategy, and volume limits.*

- **Integration requirements?** (API, Slack, email, CRM, internal tools?)  
  *Determines the delivery mechanism. The AI is just one piece of the system.*

- **Who needs to approve this?** (compliance, legal, security, leadership?)  
  *Hidden stakeholders can kill projects. Surface them early.*

- **Are there regulatory or compliance requirements?**  
  *HIPAA, GDPR, SOC 2, industry-specific rules. Non-negotiable constraints.*

- **What's the timeline?**  
  *"We need this yesterday" vs. "we're exploring for next quarter" changes your entire approach.*

## 2.3 Mapping Needs to Claude Capabilities

Once you understand what the customer needs, you need to quickly map it to the right Claude capability and technique. This table is your decision tree.

| Customer Says... | Claude Capability | Primary Technique | Model Recommendation |
|---|---|---|---|
| "We need to **categorize/classify** X" | Classification | Few-shot examples + system prompt with categories | Haiku for high-volume, Sonnet for nuanced |
| "We need to **extract** Y from Z" | Extraction | XML tags + structured output (JSON) | Sonnet for complex docs, Haiku for simple |
| "We need to **summarize** long documents" | Summarization | System prompt with format spec + length constraints | Sonnet or Opus for fidelity |
| "We need to **draft/write** responses" | Generation | System prompt + tone examples + template | Sonnet for quality, Haiku for volume |
| "We need to **answer questions** about our docs" | Q&A / RAG | Long context window OR retrieval + grounding | Sonnet or Opus (200K context) |
| "We need a **multi-step workflow**" | Chaining / Agents | Prompt chaining or tool use | Sonnet for orchestration |
| "We need **structured data** out" | Structured Output | JSON mode or tool_choice (forced function call) | Any model, depends on complexity |
| "We need to **compare/evaluate** things" | Analysis | Rubric in system prompt + structured output | Sonnet or Opus for judgment |
| "We need to **translate/transform** content" | Transformation | System prompt with style guide + examples | Sonnet for quality |
| "We need to **review/check** for errors" | Validation | Checklist in system prompt + structured feedback | Sonnet for nuance |

## 2.4 Red Flags to Listen For

During discovery, certain customer statements should trigger immediate follow-up. These aren't deal-breakers, but they require careful handling. In the interview, recognizing and addressing these signals shows maturity.

---

### "We need 100% accuracy."

**Why it's a red flag**: LLMs are probabilistic. 100% accuracy on non-trivial tasks is unrealistic.  
**How to handle it**:
- Acknowledge the aspiration: "I understand accuracy is critical. Let me dig into what kinds of errors would be most costly."
- Reframe around risk: "What happens when an error occurs? Some errors are easy to catch, others are costly."
- Propose human-in-the-loop: "What if the system handles the clear cases automatically and flags ambiguous ones for human review? That often gets you to 99%+ effective accuracy."
- Suggest confidence scoring: "We can have the model output a confidence score and only auto-process high-confidence results."

---

### "It needs to be real-time."

**Why it's a red flag**: "Real-time" means different things to different people. Sub-second? Under 5 seconds? Under a minute?  
**How to handle it**:
- Clarify: "When you say real-time, what latency would feel acceptable? Under 1 second? Under 5 seconds?"
- Discuss techniques: streaming responses, prompt caching, using Haiku for speed, pre-computing common responses
- Explore whether batch is actually fine: "Does it need to be instant, or could processing in the background with a notification work?"

---

### "We have sensitive data."

**Why it's a red flag**: PII, PHI, financial data, and classified information all have different handling requirements.  
**How to handle it**:
- Get specific: "What kind of sensitive data? Personal information? Health records? Financial?"
- Discuss Anthropic's data handling: zero data retention policy on API, SOC 2 compliance, no training on API data
- Consider architecture: Can you strip PII before sending to Claude? Do you need on-prem?
- Compliance check: "Are there specific regulations you need to comply with? HIPAA? GDPR?"

---

### "We want it to make decisions autonomously."

**Why it's a red flag**: Autonomous AI decision-making is high-risk territory, especially for consequential decisions.  
**How to handle it**:
- Assess impact: "What are the consequences of a wrong decision? Is it reversible?"
- Propose graduated autonomy: "Start with AI-assisted (human decides), move to AI-driven (human approves), then consider AI-autonomous for low-risk decisions."
- Discuss guardrails: output validation, confidence thresholds, audit logging, escalation paths
- Reference Anthropic's responsible AI principles (this shows alignment with the company)

---

### "It should work exactly like our current system."

**Why it's a red flag**: AI is not a deterministic rules engine. Trying to replicate exact behavior is fighting the technology.  
**How to handle it**:
- Validate the concern: "I want to make sure we preserve what works well about your current system."
- Reframe: "Rather than replicating the exact process, let's focus on the outcomes. What results does your current system produce that you want to preserve?"
- Address non-determinism: "AI systems are probabilistic -- they might phrase things differently each time, but can be very consistent in the decisions and classifications they make."
- Suggest testing: "We can build a test suite from your current system's outputs and validate that the AI matches on the cases that matter."

---

### "We need it by tomorrow / this week."

**Why it's a red flag**: Urgency often means corners get cut, expectations aren't set, and the project fails.  
**How to handle it**:
- Separate MVP from production: "I can build a working prototype quickly. Production-ready with error handling, monitoring, and edge case coverage takes longer."
- Propose phased rollout: "Phase 1: Working demo this week. Phase 2: Handle edge cases next week. Phase 3: Production deployment with monitoring."
- Manage expectations: "Let's identify the 20% of the problem that delivers 80% of the value and start there."

````python
def create_customer_persona(scenario_description: str) -> str:
    """Create a system prompt that turns Claude into a customer for discovery practice."""
    return dedent(f"""\
        You are role-playing as a customer in a discovery meeting with an AI solutions engineer.
        
        Your scenario:
        {scenario_description}
        
        IMPORTANT INSTRUCTIONS FOR YOUR ROLE-PLAY:
        - Stay in character as the customer at all times
        - Start with a vague, high-level description of your problem (don't volunteer details)
        - Only reveal details when the engineer asks the RIGHT questions
        - If they ask a great question, give a detailed, helpful answer
        - If they ask a vague question, give a vague answer
        - If they jump straight to a solution without asking enough questions, gently redirect:
          "That's interesting, but I'm not sure that addresses our main concern..."
        - Have some hidden constraints that only come out through good questioning:
          * A compliance requirement they haven't mentioned
          * A failed past attempt at solving this
          * A budget or timeline constraint
          * An edge case that breaks simple solutions
        - Be realistic -- you're a busy professional who wants help but isn't technical
        - If they ask to see examples, provide realistic sample data
        
        Start the conversation with your opening statement about the problem.
        Keep responses concise (2-4 sentences) unless they ask for detail.
    """)


# Example customer personas for practice
SCENARIOS = {
    "healthcare_triage": """
        You are the Director of Patient Services at a mid-size community health clinic.
        You have 3 clinics with 15 providers. Your front desk staff spend 2+ hours per day 
        reading patient messages from the patient portal and deciding urgency level and 
        which provider should respond. 
        
        Hidden details to reveal only when asked:
        - You're HIPAA-regulated (critical constraint)
        - You tried a rules-based system last year; it was too rigid and missed nuance
        - Some messages contain mental health crisis language that needs immediate escalation
        - Volume: ~200 messages/day across all clinics
        - You use Epic as your EHR system
        - Budget: you have grant funding but it expires in 6 months
        - Staff are not technical and resistant to change
    """,
    
    "ecommerce_reviews": """
        You are the Head of Customer Experience at an online furniture retailer.
        You get 1,000+ product reviews per week and your team manually reads them 
        to identify product quality issues, shipping complaints, and customer praise.
        
        Hidden details to reveal only when asked:
        - Reviews come in multiple languages (English, Spanish, French)
        - Some reviews contain photos that are important context (you only care about text for now)
        - You need to feed insights to 3 different teams: Product, Logistics, and Marketing
        - You tried sentiment analysis before but it was too simplistic
        - You have a Shopify integration requirement
        - Some reviews are fake/spam and need to be flagged
        - You want weekly summary reports, not real-time processing
    """,
    
    "legal_contracts": """
        You are a Senior Paralegal at a 50-person law firm specializing in commercial real estate.
        Your team reviews 30-40 commercial lease agreements per week, looking for non-standard 
        clauses, missing provisions, and potential risks.
        
        Hidden details to reveal only when asked:
        - Each lease is 40-80 pages long
        - You have a checklist of 25 standard provisions that must be present
        - The attorneys care most about deviation from standard language, not the standard parts
        - Confidentiality is critical -- these are client documents under attorney-client privilege
        - You've been burned by AI tools that hallucinated clause language
        - Output needs to be a structured report that goes to the reviewing attorney
        - Some leases are scanned PDFs (OCR quality varies)
    """
}

print("Available scenarios: healthcare_triage, ecommerce_reviews, legal_contracts")
print("Use the practice_discovery() function below to start a session.")
````

````python
def practice_discovery(scenario_key: str):
    """Start a single-turn discovery practice session.
    
    The customer gives their opening statement, then you provide
    your first set of discovery questions.
    """
    if scenario_key not in SCENARIOS:
        print(f"Unknown scenario. Choose from: {list(SCENARIOS.keys())}")
        return
    
    system_prompt = create_customer_persona(SCENARIOS[scenario_key])
    
    # Get the customer's opening statement
    response = client.messages.create(
        model=MODEL,
        max_tokens=500,
        system=system_prompt,
        messages=[
            {"role": "user", "content": "Hi, thanks for meeting with me today. I understand you're looking for help with a workflow challenge. Can you tell me a bit about what's going on?"}
        ]
    )
    
    customer_statement = response.content[0].text
    print("=" * 70)
    print("CUSTOMER OPENING STATEMENT")
    print("=" * 70)
    print(f"\n{customer_statement}\n")
    print("=" * 70)
    print("\nNow write your discovery questions below and use")
    print("ask_customer() to continue the conversation.")
    print("=" * 70)
    
    return system_prompt, customer_statement


# Example: Start a practice session
# session = practice_discovery("healthcare_triage")
````

````python
def ask_customer(system_prompt: str, conversation_history: list, your_questions: str):
    """Continue the discovery conversation with the customer.
    
    Args:
        system_prompt: The system prompt from practice_discovery()
        conversation_history: List of message dicts [{"role": ..., "content": ...}]
        your_questions: Your discovery questions as a string
    
    Returns:
        Updated conversation history with the customer's response
    """
    # Add your questions to the history
    updated_history = conversation_history + [
        {"role": "user", "content": your_questions}
    ]
    
    response = client.messages.create(
        model=MODEL,
        max_tokens=1000,
        system=system_prompt,
        messages=updated_history
    )
    
    customer_response = response.content[0].text
    
    # Add customer response to history
    updated_history.append({"role": "assistant", "content": customer_response})
    
    print("=" * 70)
    print("CUSTOMER RESPONSE")
    print("=" * 70)
    print(f"\n{customer_response}\n")
    
    return updated_history


# Example usage:
# system_prompt, opening = practice_discovery("healthcare_triage")
#
# history = [
#     {"role": "user", "content": "Thanks for meeting with me. Can you tell me about the challenge?"},
#     {"role": "assistant", "content": opening}
# ]
#
# history = ask_customer(system_prompt, history, """
#     Thanks for that overview. I have a few questions to make sure I understand:
#     1. Can you walk me through what happens when a patient message comes in? Step by step?
#     2. How do your staff currently decide the urgency level?
#     3. What happens when a message gets routed to the wrong provider?
# """)
````

## Full Discovery Simulation

This cell runs a complete multi-turn discovery conversation. You can study the exchange to see how good discovery unfolds.

````python
def run_full_discovery_simulation(scenario_key: str, num_turns: int = 4):
    """
    Run a full simulated discovery conversation between an AI solutions engineer
    and a customer. Useful for studying how discovery should flow.
    """
    customer_system = create_customer_persona(SCENARIOS[scenario_key])
    
    engineer_system = dedent("""\
        You are an expert AI solutions engineer at Anthropic conducting a discovery 
        meeting with a potential customer. Your goal is to understand their problem 
        deeply before proposing any solution.
        
        Follow the SPIN framework:
        - Start with Situation questions (understand the landscape)
        - Move to Problem questions (find the pain)
        - Explore Implications (understand the cost)
        - End with Need-payoff (anchor on outcomes)
        
        Rules:
        - Ask 2-3 questions per turn (not more)
        - Ask for concrete examples early
        - Listen for red flags (compliance, unrealistic expectations)
        - Do NOT propose a solution until you have enough information
        - When you do propose, map to specific Claude capabilities
        - Confirm your understanding: "So what I'm hearing is..."
        
        Keep responses concise and professional.
    """)
    
    # Initialize
    customer_messages = []
    engineer_messages = []
    
    # Engineer's opening
    engineer_opening = "Hi, thanks for taking the time to meet with me today. I'd love to understand what challenge you're facing. Can you walk me through the situation?"
    
    print("=" * 70)
    print(f"FULL DISCOVERY SIMULATION: {scenario_key}")
    print("=" * 70)
    print(f"\n{'ENGINEER':>10}: {engineer_opening}\n")
    
    customer_messages.append({"role": "user", "content": engineer_opening})
    
    for turn in range(num_turns):
        # Customer responds
        customer_response = client.messages.create(
            model=MODEL,
            max_tokens=600,
            system=customer_system,
            messages=customer_messages
        )
        customer_text = customer_response.content[0].text
        customer_messages.append({"role": "assistant", "content": customer_text})
        print(f"{'CUSTOMER':>10}: {customer_text}\n")
        
        # Engineer responds
        engineer_messages.append({"role": "user", "content": customer_text})
        
        # On the last turn, ask the engineer to summarize understanding
        if turn == num_turns - 1:
            engineer_messages_with_instruction = engineer_messages.copy()
            last_msg = engineer_messages_with_instruction[-1]
            engineer_messages_with_instruction[-1] = {
                "role": "user",
                "content": last_msg["content"] + "\n\n[INSTRUCTION: This is your last turn. Summarize your understanding and propose a high-level solution approach using Claude.]"
            }
            engineer_response = client.messages.create(
                model=MODEL,
                max_tokens=800,
                system=engineer_system,
                messages=engineer_messages_with_instruction
            )
        else:
            engineer_response = client.messages.create(
                model=MODEL,
                max_tokens=500,
                system=engineer_system,
                messages=engineer_messages
            )
        
        engineer_text = engineer_response.content[0].text
        engineer_messages.append({"role": "assistant", "content": engineer_text})
        customer_messages.append({"role": "user", "content": engineer_text})
        print(f"{'ENGINEER':>10}: {engineer_text}\n")
        print("-" * 70)
    
    print("\n" + "=" * 70)
    print("END OF SIMULATION")
    print("=" * 70)


# Uncomment to run:
# run_full_discovery_simulation("healthcare_triage")
# run_full_discovery_simulation("ecommerce_reviews")
# run_full_discovery_simulation("legal_contracts")
````

## 2.6 The First 10 Minutes Checklist

In the customer scenarios interview, you typically have about 10 minutes for discovery before you should start building. Here's what you need to accomplish in that time.

### Must-Have (Non-Negotiable)

- [ ] **Understand the org and their mission** -- who are they and what do they do?
- [ ] **Understand the specific task/workflow** -- what exactly is the process you're improving?
- [ ] **See an example of input AND desired output** -- the single most important request you can make
- [ ] **Understand volume and latency needs** -- 10/day vs. 10,000/day changes everything
- [ ] **Understand success criteria** -- how will they know if this works?
- [ ] **Identify the top edge cases** -- what makes this hard?
- [ ] **Identify hard constraints** -- PII, compliance, budget, timeline
- [ ] **Confirm understanding** -- "So what I'm hearing is... does that sound right?"

### Should-Have (Ask If Time Permits)

- [ ] **Understand the integration landscape** -- where does this fit in their tech stack?
- [ ] **Understand stakeholders** -- who else needs to approve or use this?
- [ ] **Understand past attempts** -- what have they tried before and why did it fail?
- [ ] **Understand the human fallback** -- what happens when the AI can't handle something?

### Transition Statement

When you're ready to move from discovery to building, use a clear transition:

> "OK, let me make sure I have this right. You need [task] that takes [input] and produces [output]. The volume is [X], it needs to integrate with [Y], and the key constraint is [Z]. Your success metric is [metric]. Does that capture it? ...Great. Let me start building something and we can iterate."

This does three things:
1. Confirms your understanding (catches misunderstandings before you build)
2. Shows the interviewers you were listening
3. Creates a clean break between discovery and execution

## Quick-Reference: Discovery Anti-Patterns

Things that will cost you points in the interview:

| Anti-Pattern | What It Looks Like | Why It's Bad |
|---|---|---|
| **Premature solutioning** | "Oh, you should use RAG for that." (after 30 seconds) | You don't understand the problem yet |
| **Interrogation mode** | Rapid-fire questions with no acknowledgment | Feels like a checklist, not a conversation |
| **Tech-first thinking** | "What model should we use?" (before understanding the task) | Puts the tool before the problem |
| **Assuming the problem** | "So you need a chatbot." (they never said that) | Shows you're not listening |
| **Ignoring constraints** | Building a solution that requires PII in the API without asking | Shows naivety about real-world deployment |
| **Not asking for examples** | Designing a system without ever seeing real input/output | You're guessing, not engineering |
| **Over-engineering** | Proposing a multi-agent system for a classification task | Complexity is not a virtue |

## Summary

### Key Takeaways

1. **Discovery is 50% of the interview.** Bad discovery leads to building the wrong solution. No amount of technical brilliance compensates for solving the wrong problem.

2. **The best candidates LISTEN more than they talk.** In the first 10 minutes, your ratio should be roughly 30% talking (questions) and 70% listening (understanding answers, asking follow-ups).

3. **Always confirm understanding before building.** The transition statement ("So what I'm hearing is...") is not optional. It's the bridge between discovery and execution, and it's the moment where misunderstandings get caught.

4. **Ask for examples immediately.** "Can you show me what the input looks like and what the ideal output would be?" is the single highest-value question you can ask. Everything else is context for interpreting those examples.

5. **Map needs to capabilities, not the other way around.** Start with the customer's problem, then find the Claude capability that fits -- not the other way around. If you start with "I want to use tool_use," you're doing it backwards.

6. **Red flags are opportunities, not obstacles.** When a customer says something concerning ("we need 100% accuracy"), the right response is curiosity, not dismissal. Dig into why they feel that way and address the underlying concern.
